# CreaseIQ — Exploratory data analysis

A thin notebook: every computation lives in `src/creaseiq` (tested). This notebook only calls it and plots.
Run `creaseiq validate` first to build the processed cache.

```python
from creaseiq.config import get_settings
from creaseiq.data.pipeline import load_processed
from creaseiq.analytics import toss_analysis, team_stats, venue_stats, season_trends
from creaseiq.viz import charts

matches, players = load_processed(get_settings())
matches.shape
```

## The team1/team2 trap
Before 2018 `team1` is not the side batting first, so raw order must never be a feature.

```python
matches.groupby('season_year')['team1_bats_first'].mean().round(2)
```

## Toss and chasing

```python
toss = toss_analysis.toss_effect(matches)
chase = toss_analysis.chasing_advantage(matches)
toss['overall'], chase['overall']
```

```python
charts.chase_rate_by_season(chase['by_season'])
```

## Scoring trends and the Impact Player era

```python
charts.scoring_trend(season_trends.scoring_by_season(matches))
```
